Tasks 07, 08 and 11 use the GitHub CLI (`gh`) from lecture 06. Log in once before you start:

1. Run `gh auth login`.
2. Choose `GitHub.com`, then `HTTPS`, and answer `Y` when asked to authenticate Git with your GitHub credentials.
