### Running shell commands in a notebook

Jupyter can run shell commands directly. The `!` prefix runs one line:

````python
!ls -lh
````

The `%%bash` magic runs the whole cell as a bash script, so several commands can go together:

````python
%%bash
pwd
ls
````

One caveat: every cell starts in the notebook's own directory, so a `cd` in one cell does not carry over to the next. Keep this notebook in the folder where you create `penguin-survey` (Task 01), and put the `cd` and the command in the same `%%bash` cell, for example `cd penguin-survey` followed by `ls data/raw`. Use `%%bash` rather than `!` for brace expansion such as `{1..5}`, which does not work with `!` on Ubuntu and WSL.
